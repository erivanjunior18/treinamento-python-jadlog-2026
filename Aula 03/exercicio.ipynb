<h5 style="color:#2dede4;">
  💰 Projeto – Controle de Transações Financeiras
</h5>

<p style="font-size: 14px">
Neste exercício, você irá desenvolver um programa em <strong>Python</strong> que simula
o controle básico de transações financeiras, semelhante ao funcionamento de um
caixa interno ou sistema simples de movimentações financeiras em uma empresa.
</p>

<p style="font-size: 14px">
O sistema permitirá registrar <strong>entradas</strong> e <strong>saídas</strong> de valores,
controlando o saldo e armazenando o histórico de transações realizadas.
</p>

<hr>

<p style="font-size: 18px; color:#2dede4">
  🎯 Objetivo
</p>
<p style="font-size: 14px">Criar um programa que:</p>
<ul style="font-size: 14px">
  <li>Controle o saldo financeiro inicial</li>
  <li>Permita registrar transações de entrada e saída</li>
  <li>Valide se há saldo suficiente para saídas</li>
  <li>Armazene o histórico das transações</li>
  <li>Exiba o resumo final ao encerrar o programa</li>
</ul>

<hr>

<p style="font-size: 18px; color:#2dede4">
  📥 Dados de Entrada
</p>
<p style="font-size: 14px">Durante a execução do programa, o usuário deverá informar:</p>
<ul style="font-size: 14px">
  <li>A opção do menu desejada</li>
  <li>O valor da transação (entrada ou saída)</li>
</ul>

<hr>

<p style="font-size: 18px; color:#2dede4">
  📋 Menu de Operações
</p>

<table border="1" cellpadding="8" cellspacing="0">
  <thead style="background-color:#e0f2f1;">

    </tr>
    <tr>
      <td>3</td>
      <td>Finalizar o programa</td>
    </tr>
  </tbody>
</table>

<br>

<p style="font-size: 18px; color:#2dede4">
  ⚙️ Regras de Funcionamento
</p>
<ul style="font-size: 14px">
  <li>O saldo inicial deve começar em <strong>R$ 0,00</strong></li>
  <li>Entradas aumentam o saldo</li>
  <li>Saídas só podem ocorrer se houver saldo suficiente</li>
  <li>Todas as transações devem ser armazenadas em uma lista</li>
</ul>

<hr>

<p style="font-size: 18px; color:#2dede4">
  📤 Saída Esperada
</p>
<p style="font-size: 14px">Ao finalizar o programa, deve ser exibido:</p>
<ul style="font-size: 14px">
  <li>O histórico completo de transações (entrada e saída)</li>
  <li>O saldo final disponível</li>
</ul>

<p style="background:#f1f8e9; padding:2px; border-left:6px solid #000000; color: #000051; font-size: 14px">

<strong>if</strong> para validações e <strong>match / case</strong> para tratar as opções do menu.
</p>
