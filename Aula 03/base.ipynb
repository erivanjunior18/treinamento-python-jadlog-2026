```python
# Simples
# def = para criar própria função
# a função só funcionará se for chamada

def dizOi():
    print ('Oi')

def soma2valores():
    print(10+20)

dizOi()
soma2valores()
```

```python
# Com retorno

def dizOi():
    return 'Oi'

def soma2valores():
    return 10+20

print(dizOi())
print(soma2valores() * 100)
```

```python
# Com parâmetros e argumentos

def cantar_parabens(nome):
    return f'Oi, {nome}! Parabens para vc lalalala lilili'

#cantar_parabens(input('Aniversariante: '))

def soma2valores(num1,num2):
    return num1 + num2

soma2valores(15,30)
```

```python
#como gerar dica da função criada

def geradordesconto(valor, desconto = 0.1, *args, **kwargs): #se o parametro de descontro estiver definido o python calculará o valor padrão, se não na chamada deverá definir o valor do desconto.
    '''
        A função calcula um desconto por % inserido

        Valor: pode ser float ou int

        Desconto: precisa ser informado em decimal
    '''

    return valor * desconto

geradordesconto(100, 0.2)
```
