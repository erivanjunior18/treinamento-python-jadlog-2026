# transações

```python
# função 'pass' serve para pular o erro.

saldo = 0
lista_transacoes = []

while True:
    menu_tipo_transacao = int(input('1-Depósito | 2-Saque | 3-Sair'))
    tentativas = 0

    if menu_tipo_transacao != 3:
        valor = float(input('Qual valor?'))
    
    match menu_tipo_transacao:
        case 1:
            print(f'Depósito Efetuado no valor de R${valor}')
            saldo += valor
            #lista_transacoes.append(f'DEP - {valor}')
            lista_transacoes.append(valor)

        case 2:
            if valor <= saldo: 
                saldo -= valor
                print(f'Saque Efetivado no valor de -R${valor}')
                print(f'Saldo atual R$ {saldo}')
                lista_transacoes.append(valor * -1)
            else:
                print(f'Saldo insuficiente. Saldo atual: R${valor}')
        case 3:
            break
        case _:
            pass

print(lista_transacoes)
print(saldo)
```
