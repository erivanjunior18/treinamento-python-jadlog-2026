```python
#Atalho para duplicar as linhas ALT+SHIFT+SETA PARA BAIXO
import pandas as pd

dados = [
    ['Ana', 'SP',1800],
    ['Joao', 'RJ',2500],
    ['Maria', 'DF',9800],
    ['Isabel', 'PB',35800],
    ['Luciano', 'MT',9800],
    ['José', 'AM',8800],
]

colunas = ['Nome','Estado', 'Salario']

#dataframe nome chique para tabela
df = pd.DataFrame(dados, columns=colunas)

df.head(2)  # 5 primeiras e 5 últimas
df.shape    # qtd de linhas e colunas (sem o parenteses)
df.info     # sempre rodar antes de comecar a analise. Retornará o tipo de dados das colunas
df.describe() # retornar calculo simples
df.dropna() # remover colunas vazias
df['Nome']
```

```python
df.to_excel('dados_bobos.xlsx', index=False)
```

```python
colunas_exercicio = ['Comprador','Produto', 'Valor','Qtd']

dados_exercicio = [
    ['Ana', 'TV',1800,2],
    ['Joao', 'Smartphone',2500,6],
    ['Maria', 'Smartphone',9800,3],
    ['Isabel', 'TV',35800,5],
    ['Luciano', 'Geladeira',9800,6],
    ['José', 'Fogão',8800,4],
]


#dataframe nome chique para tabela
df_exercicio = pd.DataFrame(dados_exercicio, columns=colunas_exercicio)
```

```python
df_exercicio[ df_exercicio['Valor'] > 5000]

df_exercicio.describe()

df_exercicio['Valor'].max()

df_exercicio.groupby('Produto')['Valor'].max()

df_exercicio.groupby('Produto').agg(
    maior_preco = ('Valor', 'max'),
    maior_venda = ('Qtd','max'),
    total_venda = ('Valor','sum')
)

```
