Match/Case

```python
menu = int(input("1-Cadastrar | 2-Visualizar | Opção: "))

match menu: 

    case 1:
        print("escolheu 1")
    case 2:
        print("escolheu 2")

    case _:
        print('Resposta inválida!')
```
