```python
nome = str(input("Nome do funcionário: "))
cargo = str(input("Cargo: 1-Analista Financeiro, 2-Assistente Financeiro, 3-Coordenador"))
salario = float(input("Salário: "))
tempo_casa = int(input("Anos de experiência: "))


print(nome)
print(cargo)
print(salario)
print(tempo_casa)

```

```python
#regra match

match cargo:
    case 1: "Analista Financeiro"
    case 2: "Assistente Financeiro"
    case 3: "Coordenador"
    case _: "Opção Inválida!"
```

```python
#Regra de Classificação - Senioridade
if tempo_casa <= 2 & tempo_casa >=0:
    senioridade = 'Júnior'
elif tempo_casa <=5 & tempo_casa >= 3:
    senioridade = 'Pleno'
elif tempo_casa >= 6 & tempo_casa >=40:
    senioridade = 'Sênior'
else: 
    senioridade = 'erro'
```

```python
#aplicação de logica de decisao

if cargo == '1' and senioridade != 'Júnior':
    permissao = "Gerenciamento liberado"

elif cargo == '2' and senioridade != 'Júnior':
    permissao = "Apenas operações básicas"

elif cargo == '3' and senioridade != 'Sênior':
    permissao = "Acesso total"

elif cargo == 'erro':
    permissao = "Verifique as respostas anteriores"

else:
    permissao = 'Permissão negada!'
    
permissao
```

```python
# Impressão

print(f"Nome: {nome}\nCargo: {cargo}\ntempo_casa: {tempo_casa}\nsalario: {salario}\n\
senioridade: {senioridade}\npermissao: {permissao}")
```
