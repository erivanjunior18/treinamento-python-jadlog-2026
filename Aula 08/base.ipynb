Importar bibliotecas

```python
import pandas as pd
```

Criando dataframe

```python
#Tabela criada no python
df_entregas = pd.DataFrame({
        'id_entrega': [1,2,3], #chave primária
        'filial'    : ['FL RIO', 'FL VIX', 'FL REC'],
        'status'    : ['ENTREGUE', 'ATRASADO', 'ENTREGUE']
})

df_custos = pd.DataFrame({
        'id_custo'      : [1,2,3],
        'id_entrega'    : [1,1,3],          # chave estrangeira
        'custo'         : [114.74,89.99,35.77]
})

```

JOIN

```python
# inner join

pd.merge(
    df_entregas, df_custos,
    on = 'id_entrega',
    how= 'inner'    # apenas as linhas onde há informação
)
```

```python
# full outer join

pd.merge(
    df_entregas, df_custos,
    on = 'id_entrega',
    how= 'outer'    
)
```

```python
# left join

pd.merge(
    df_entregas, df_custos,
    on = 'id_entrega',
    how= 'left'    
)
```

```python
# right join

pd.merge(
    df_entregas, df_custos,
    on = 'id_entrega',
    how= 'right'    
)
```

```python
df_entregas_idx = df_entregas.set_index('id_entrega')
df_custos_idx = df_custos.set_index('id_entrega')

df_join = df_entregas_idx.join(df_custos_idx, how='left')
df_join.reset_index()
```
