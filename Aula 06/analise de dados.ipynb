```python
nome_arquivo = 'base_entregas_brutas'

def ler_arquivo(nome_arquivo):
    try:
        df = pd.read_excel(f'dados/{nome_arquivo}.xlsx')
        return df

    except FileNotFoundError as e:
        print(f'Erro: {e}')

def criar_atualizar_arquivo(nome_arquivo, df_tratado=''):
    df_tratado.to_excel(f'dados/{nome_arquivo}.xlsx'
                        )
```

```python
df = ler_arquivo('base_entregas_brutas')
```

```python
df_tratado = df.copy()
```

<p style="font-size: 1.1rem; color: #00d9ff; font-family: monospace ">
    <strong>
        Filial 
    </strong>
</p>

```python
#pegando todas a linhas, mas não sabe qual coluna deve informar qual coluna quer
df_tratado.loc[: , 'filial'] = df_tratado['filial'].fillna('Não Informada') #vazio
#df.replace() para substituir alguma informação

df_tratado['filial'].sample(10)
```

<p style="font-size: 1.1rem; color: #00d9ff; font-family: monospace ">
    <strong>
        Cliente 
    </strong>
</p>

```python
df_tratado.loc[: , 'cliente'] = df_tratado['cliente'].fillna('Outros') #vazio
```

<p style="font-size: 1.1rem; color: #00d9ff; font-family: monospace ">
    <strong>
        Dias de atraso 
    </strong>
</p>

```python
df_tratado.loc[: , 'dias_atraso'] = df_tratado['dias_atraso'].fillna(0) #vazio
```

<p style="font-size: 1.1rem; color: #00d9ff; font-family: monospace ">
    <strong>
        Prazo previsto 
    </strong>
</p>

```python
df_tratado.loc[: , 'prazo_previsto'] = df_tratado['prazo_previsto'].fillna(
    df_tratado['prazo_previsto'].median()
)
```

<p style="font-size: 1.1rem; color:#9beb3b; text-align: center; font-family: monospace ">
    <strong><br/><br/><br/><br/><br/><br/>
        Tratamento - Duplicadas

</p>

<p style="font-size: 1.1rem; color: #00d9ff; font-family: monospace ">

```python
df_tratado.duplicated(subset=['id_entrega']).sum()
```

<p style="font-size: 1.1rem; color: #00d9ff; font-family: monospace ">
    <strong>
        Removendo duplicadas, mantendo a primeira ocorrencia
    </strong>
</p>

```python
df_tratado.drop_duplicates(subset=['id_entrega'],inplace = True)

#df_tratado.drop_duplicates(subset=['id_entrega']
```

<p style="font-size: 1.1rem; color: #00d9ff; font-family: monospace ">
    <strong>

</p>

<p style="font-size: 1.1rem; color:#9beb3b; text-align: center; font-family: monospace ">
    <strong><br/><br/><br/><br/><br/><br/>
        Tratamento - Tipagem correta para ID
    </strong>
</p>

```python
df_tratado['id_entrega'] = df_tratado['id_entrega'].astype('Int64')
df_tratado['prazo_previsto'] = df_tratado['prazo_previsto'].astype('Int64')
df_tratado['dias_atraso'] = df_tratado['dias_atraso'].astype('Int64')
```

```python
df_tratado.dtypes #retorna os tipos de dados
```

<p style="font-size: 1.1rem; color:#9beb3b; text-align: center; font-family: monospace ">
    <strong><br/><br/><br/><br/><br/><br/>
        Tratamento - Padronização e Normalização
    </strong>
</p>

<p style="font-size: 1.1rem; color: #00d9ff; font-family: monospace ">
    <strong>
        Removendo espaços e alterando para caixa alta 
    </strong>
</p>

```python
df_tratado['status'] = df_tratado['status'].str.strip().str.upper() #remover espaço antes e no final do texto da celula

df_tratado['tipo_servico'] = df_tratado['tipo_servico'].str.strip().str.upper() #remover espaço antes e no final do texto da celula

df_tratado['cliente'] = df_tratado['cliente'].str.strip().str.upper() #remover espaço antes e no final do texto da celula

df_tratado.head()
```

<p style="font-size: 1.1rem; color: #00d9ff; font-family: monospace ">
    <strong>
        Status 
    </strong>
</p>

```python
#chaves: como está : como ficará
mapa_status = {
    'EM TRÂNSITO' : 'EM TRANSITO',
    'EM TRANSITO' : 'EM TRANSITO',
    'ENTREGUE'    : 'ENTREGUE',
    'EM TRÂNZITO' : 'EM TRANSITO',
    'EM TRANZITO' : 'EM TRANSITO',
    'ATRAZADO'    : 'ATRASADO',
    'ATRASADO'    : 'ATRASADO'
}

df_tratado['status'] = df_tratado['status'].map(mapa_status)

df_tratado.sample(frac=0.15) #frac retorna uma porcentagem de linhas para amostra
```

Criando coluna calculada

```python
df_tratado['data_prevista'] = (
    df_tratado['data_postagem'] +
    pd.to_timedelta(df_tratado['prazo_previsto'],unit='D') # sempre D para dias | M mês | Y ano >> Sempre tem que converter em uma unidade de tempo
)

df_tratado[[
    'data_postagem',
    'prazo_previsto',
    'data_prevista'
]
]
```

```python
hoje =pd.Timestamp.today().normalize()

df_tratado['Atraso'] = (
    hoje - df_tratado['data_prevista']
).dt.days

df_tratado['Atraso'] = df_tratado['Atraso'].clip(lower=0)
```

<p style="font-size: 1.1rem; color:#9beb3b; text-align: center; font-family: monospace ">
    <strong><br/><br/><br/><br/><br/><br/>
        Prazo previsto de acordo com a filial
    </strong>
</p>

```python
#ALT+SHIFT+SETA PARA CIMA para duplicar as celulas
#Após salvar a celula basta clicar na letra B para abrir uma nova celula

df_tratado.loc[
    (df_tratado['cliente'] == "SHOPEE") &
    (df_tratado['filial'] == "RJ") ,
    'prazo_previsto'
] = 5

#df_tratado.loc[
 #   (df_tratado['cliente'] == "AMAZON") &
  #  (df_tratado['filial'].isin(["RJ", 'SP'])) ,
  #  'prazo_previsto'
#] = 5


df_tratado.loc[
    (df_tratado['cliente'] == "AMAZON") &
    (df_tratado['filial'] .isin(["MG", 'SP'])) ,
    'prazo_previsto'
] = 2
```

<p style="font-size: 1.1rem; color:#9beb3b; text-align: center; font-family: monospace ">
    <strong><br/><br/><br/><br/><br/><br/>
        Atualizando arquivo para executar fórmulas
    </strong>
</p>

```python
#criar uma instancia que vai abrir o arquivo e a visibilidade vai mostrar se o excel abriu
app = xw.App(visible=False)

#qual arquivo vai abrir - Informar sempre o caminho completo
wb = xw.Book(f'dados/base2.xlsx')

#forçar os calculos da planilha
wb.app.calculate()

#salvar o excel
wb.save()

#fechar o excel
wb.close()

#encerra a programação
app.quit()
```
