```python
#for variável in alguma coisa:
#range (início, fim, passo)
#com o for é necessário informar quantas vezes a informação deve ser repetida

for contador in range(10,-1,-1):
    if contador % 2 == 1:
        print(contador, end=" ")
```

<p style="font-size: 1.1rem; color:#9beb3b; text-align: center; font-family: monospace ">
    <strong></br></br></br></br>
        Validando um senha de brincadeira
    </strong>
</p>

```python
senha_cadastrada = input('Senha: ')
senha_entrada = ''
tentativas = 0

while senha_cadastrada != senha_entrada:
    senha_entrada = input("Digite sua senha: ")
    tentativas += 1

    if tentativas == 3 and senha_cadastrada != senha_entrada:
        print("Tentativas máximas atingidas")
        break

if senha_cadastrada == senha_entrada:
    print("Acesso Permitido")
```

```python
# Entrada de valores


franquias=['RJ Rio de Janeiro 55', 'LJ São Paulo 55', 'LJ Tres Rios 01']

fat_franquia = []

for franquia in franquias:
    valor = float(input(f'{franquia} | Valor: '))

    fat_franquia.append(franquia)
    fat_franquia.append(valor)

fat_franquia
```

```python
# Cálculo

#quando o calculo estiver com indice, o calculo é feito a nivel de linha
soma = 0
for indice, dado in enumerate(fat_franquia):
    if indice % 2 == 1:
        soma += dado * 0.1
       # print(dado * 0.1)

print(soma)
```

```python
#alt + seleção do mouse adiciona valores em várias linhas de uma só vez
#crtl + ; coloca/tira a linha como comentário

franquias_cadastradas = []

for contador in range(2):

    franquia = dict(

        id      = int(input('Código: ')),
        cnpj    = input('CNPJ: '),
        endereco= input('Endereço: ').title(), #title = insere a primeira letra maiuscula de cada palavra
        bairro  = input('Bairro: ').title(),
        cidade  = input('Cidade: ').title(),
        estado  = input('Estado: ').upper(),
    )
    franquias_cadastradas.append(franquia)
```

```python
for indice, franquia in enumerate(franquias_cadastradas):
    print(f'{indice + 1}ª Franquia\n')

    for chave, valor in franquia.items():
        print(f'{chave:<9}: {valor}')
    print()
```
